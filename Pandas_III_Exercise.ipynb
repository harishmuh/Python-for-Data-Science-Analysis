```python
audi_url = 'https://raw.githubusercontent.com/harishmuh/Python-for-Data-Science-Analysis/refs/heads/main/datasets/audi.csv'
df_audi = pd.read_csv(audi_url)
df_audi.head()
```

```python
toyota_url = 'https://raw.githubusercontent.com/harishmuh/Python-for-Data-Science-Analysis/refs/heads/main/datasets/toyota.csv'
df_toyota = pd.read_csv(toyota_url)
df_toyota.head()
```

## **No.1**

- If you notice, the column name 'tax(£)' in the Hyundai dataset is different from the other datasets.

## **No.8**

Create a dataframe containing three columns:
- Column Name: Column name
- Number of Uniques: Number of unique values
- Unique Values: Unique data/values

```python
# Method 1 (one by one)
number_of_unique = df_car.nunique()
number_of_unique
```

```python
uniques = df_car.apply(lambda x: x.unique())
uniques
```

```python
pd.DataFrame({
    'Column Name' : number_of_unique.index,
    'Number of Unique' : number_of_unique.values,
    'Unique Values' : uniques
})
```

```python
# Method 2 (For Loop)
list_unik = []

for kolom in df_car:
    list_unik.append([kolom, df_car[kolom].nunique(), df_car[kolom].unique()])

pd.DataFrame(list_unik, columns=['Column Name', 'Number of Unique', 'Unique Values'])
```

## **No.17**

Show cars whose model names contain the letter 'x'!

```python
# method 1
df_car[df_car['model'].str.contains('x|X')]['model'].unique()
```

```python
# Method 2: Using apply

def check_x(name):
  if 'x' in name.lower():
    return True
  else:
    return False
```

```python
df_car[df_car['model'].apply(check_x)]['model'].unique()
```

## **No.18**

Create a new dataframe with a pivot table containing the average tax for each transmission and fuel type.

```python
df_car.pivot_table(
    index = 'transmission',
    columns = 'fuelType',
    values = 'tax',
    aggfunc = 'mean'
).fillna(0)
```

## **No.19**

Create a new dataframe with a pivot table containing the average price for each brand and transmission with a year column.

```python
df_car.pivot_table(
    index = ['brand', 'transmission'],
    columns = 'year',
    values = 'price',
    aggfunc = 'mean'
).fillna(0).round(0)
```
