# **REST API Tutorial: Get Data from PokeAPI**

REST API: representational state transfer (REST) - Application Programming Interface.

We will demonstrate how to REST API data to get pokemon data from PokeAPI.

### **Install and import requests**

```python
# Install or import
# Install requests if not installed yet
#!pip install requests

import requests
```

### **Define the API Endpoint**

The base URL of PokeAPI: https://pokeapi.co/api/v2/


We can try to find a pokemon data. For this instance, we will use pikachu's data.


To get Pikachu's data, the endpoint is:
https://pokeapi.co/api/v2/pokemon/pikachu


### **Send a get request**

```python
# get request
url = "https://pokeapi.co/api/v2/pokemon/pikachu"
response = requests.get(url)

# Check status code
print("Status Code:", response.status_code)

# Convert response to JSON # 200 means successful
data = response.json()
```

### **Explore Pikachu's Data**

```python
# Show all available keys
print("Available keys:", data.keys())
```

### **Extract Useful Information**

```python
# Pokemon name & ID
print("Name:", data["name"])
print("ID:", data["id"])
```

```python
# Pokemon height & weight
print("Height:", data["height"])
print("Weight:", data["weight"])
```

```python
# Pokemon's type
print("Types:")
for t in data["types"]:
    print("-", t["type"]["name"])
```

```python
# Pokemenon's stats (HP, attack, etc)
print("Stats:")
for stat in data["stats"]:
    print(f"{stat['stat']['name']}: {stat['base_stat']}")
```

```python
# Sprite Image (image url)
print("Sprite URL:", data["sprites"]["front_default"])
```

We can do to other pokemons as well. To simplify the process, we will set the base url and use variable of pokemon_name that can be modified according our pokemon interest.

```python
# Setting the base url
base_url = "https://pokeapi.co/api/v2/"
```

```python
# Define function 'get pokemon info'

def get_pokemon_info(name):
    url = f"{base_url}/pokemon/{name}"
    response = requests.get(url)

    if response.status_code == 200:
        pokemon_data = response.json()
        return pokemon_data
    else:
        print(f"Failed to retrieve data {response.status_code}")
```

```python
# Let's try it
pokemon_name = "bulbasaur"
pokemon_info = get_pokemon_info(pokemon_name)

if pokemon_info:
    print(f"Name: {pokemon_info["name"].capitalize()}")
    print(f"Id: {pokemon_info["id"]}")
    print(f"Height: {pokemon_info["height"]}")
    print(f"Weight: {pokemon_info["weight"]}")
    print("Types:")
    for t in pokemon_info["types"]:
      print("-", t["type"]["name"])
    print("Stats:")
    for stat in pokemon_info["stats"]:
      print(f"{stat['stat']['name']}: {stat['base_stat']}")
```

```python
# Sprite Image (image url)
print("Sprite URL:", pokemon_info["sprites"]["front_default"])
```
