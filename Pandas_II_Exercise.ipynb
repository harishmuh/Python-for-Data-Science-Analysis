**19. Show 7 jobs with the highest frequency**

```python
# Metod 1
df['JobTitle'].value_counts().to_frame().reset_index().rename(columns={'index':'JobTitle', 'JobTitle':'count'}).head(7)
```

```python
# Method 2
dict_job = {}

for title in df['JobTitle']:
    if title not in dict_job.keys():
        dict_job[title] = 1
    else:
        dict_job[title] += 1

pd.DataFrame({
    'JobTitle' : dict_job.keys(),
    'Frekuensi' : dict_job.values()
}).sort_values('Frekuensi', ascending=False).head(7)
```
