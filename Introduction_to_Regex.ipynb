# **Introduction to Regex**

A regular expression (shortened to regex or regexp) is a sequence of characters that defines a search pattern. Such patterns are commonly used by string search algorithms to perform "search" or "search and replace" operations on strings, or to check input strings. Regular expressions are a technique developed in the fields of theoretical computer science and formal language theory.

The concept emerged in the 1950s when American mathematician Stephen Cole Kleene formalized the description of a regular language. Regular expressions are used to match or search for patterns in text such as in search engines, in search and replace dialogs in word processors and text editors, in text processing utilities and in lexical analysis. Most programming languages ​​provide regular expression processors either built-in or through libraries.

## **1. Searching for Patterns in Text**

One of the most common uses of the re module is to find patterns in text. Let's do a quick example of using the search method in the re module to find some text:

```python
# List of patterns to search for
patterns = [ 'cat', 'dog' ]

# Text to parse
text = 'This is a string with the word cat, but no other words'

for pattern in patterns:
  print(f'Searching for pattern "{pattern}" in: \n"{text}"')

# Check for match
if re.search(pattern, text):
  print('Match was found. \n')
else:
  print('No Match was found.\n')
```

Now we have seen that `re.search()` will take a pattern, scan the text, and then return a **Match** object. If no pattern is found, **None** is returned. To give you a clearer picture of this match object, look at the cell below:

```python
# List of patterns to search for
pattern = 'cat'

# Text to parse
text = 'This is a string with the word cat, but no other words'

match = re.search(pattern, text)

type(match)
```

The **Match** object returned by the `search()` method is more than just a **Boolean** or **None**, it contains information about the match, including the original input string, the regex used, and the location of the match. Let's look at the methods we can use on the match object:

```python
# Show start of match
match.start()
```

```python
# Show start of match
match.end()
```

## **2. Split with regular expressions**

Let's see how we can split with the re syntax. This will look similar to how you would use the `split()` method on a string.

```python
# Term to split on
split_term = '@'

phrase = 'What is the domain name of someone who uses the email: hello@gmail.com'

# Split the phrase
re.split(split_term,phrase)
```

## **3. Substitutes with regular expressions**

Another very useful regex function is `re.sub()`, which replaces one bit of text with another if it finds a match. Here is an example:

```python
sentence = '''The analytics firm uses a range of BI tools to visualize data. Their internal data science team suggests
bi tools may be their most valuable resource.'''

match = re.sub(r'bi','business intelligence', sentence, flags=re.IGNORECASE)
# The IGNORECASE flag allows for matches regardless of the case

print(match)
```

## **5. Character Set**

Character set is used when you want to match one of a group of characters at a point in the input. Parentheses are used to create a character set input. For example: input [ab] searches for occurrences of a or b. In general, the rules for character sets are as follows:

- using square brackets []
- [] indicates the desired character
- [] represents a slot for a single character

Let's look at some examples:

- search/find the word `cough`

```python
words = 'Orang Batak botak kepala batok pakai batik batuk-batuk.'
pattern =  'batuk'

re.findall(pattern = pattern, string = words)
# output is [ 'batuk',  'batuk']
#  'batuk' from  'batuk'
#  'batuk' from  'batuk'
```

- search/find the words `batik` and `batuk`

```python
words = 'Orang Batak botak kepala batok pakai batik batuk-batuk.'
pattern = 'bat[iu]k'

re.findall(pattern = pattern, string = words)
# only match with words whose 3rd index is the letter 'i' or 'u'
```

- search for/find the words `Batak, batok, batik` and `batuk`

```python
words = 'Orang Batak botak kepala batok pakai batik batuk-batuk.'
pattern = '[Bb]at[aiuo]k'

re.findall(pattern = pattern, string = words)
# only match with words whose 3rd index is the letter 'a', 'i', 'u' or 'o'
```

## **6. Character Range**

As character sets grow larger, typing out every character that should (or shouldn't) match can become tedious. A more concise format using character ranges allows you to specify the character set to include all adjacent characters between a start point and an end point. The format used is `[start-end]`.

A common use case is to search for a specific range of letters in the alphabet, such as `[a-f]` will match any letter between a and f.

Let's look at a few examples:

- search for words with 5 letters, starting with 'de', then followed by any 3 lower case letters.

```python
words = 'debam, de.um, deCAK, dengung, denTum, denting, der1k, derIt, dering, des4H, desing, des!r, deSI5, det@k'

pattern = 'de[abcdefghijklmnopqrstuvwxyz][abcdefghijklmnopqrstuvwxyz][abcdefghijklmnopqrstuvwxyz]'  # character set : mentions one by one the letters from a to z

re.findall(pattern = pattern, string = words)
```

```python
words = 'debam, de.um, deCAK, dengung, denTum, denting, der1k, derIt, dering, des4H, desing, des!r, deSI5, det@k'
pattern = 'de[a-z][a-z][a-z]'  # character range : specifies the range of letters

re.findall(pattern = pattern, string = words)
```

```python
words = 'debam, de.um, deCAK, dengung, denTum, denting, der1k, derIt, dering, des4H, desing, des!r, deSI5, det@k'
pattern = 'de[a-zA-Z][a-zA-Z][a-zA-Z]'

re.findall(pattern = pattern, string = words)
```

```python
words = 'debam, de.um, deCAK, dengung, denTum, denting, der1k, derIt, dering, des4H, desing, des!r, deSI5, det@k'
pattern = 'de[a-zA-Z0-9!-_][a-zA-Z0-9!-_][a-zA-Z0-9!-_]'

re.findall(pattern = pattern, string = words)
```

## **7. Quantified Repetition**

Sometimes we want to determine the number of repetitions we want without having to rewrite the desired characters many times. We can use the syntax `{minimum_characters, maximum_characters}` where `minimum_characters` is mandatory. The syntax will be attached to the character before the `{}` sign.

- create a pattern that matches `XRA 000, 1AA 2CD`, and `XYZ 481`

```python
words = 'XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, XR1 000, ABC 77, COD 1, XY@ 123'
pattern = '[A-Z0-9][A-Z]{2} [0-9][A-Z0-9]{2}'

re.findall(pattern = pattern, string = words)

# pattern '[A-Z0-9][A-Z]{2} [0-9][A-Z0-9]{2}' is equivalent to
# pattern '[A-Z0-9][A-Z][A-Z] [0-9][A-Z0-9][A-Z0-9]'
```

- create a pattern that matches `XYZ 4810`

```python
words = 'XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, XR1 000, ABC 77, COD 1, XY@ 123'
pattern = '[A-Z0-9][A-Z]{2} [0-9][A-Z0-9]{3}'

re.findall(pattern = pattern, string = words)
```

- create a pattern that matches `XRA 000, 1AA 2CD` and `XYZ 4810`

```python
words = 'XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, XR1 000, ABC 77, COD 1, XY@ 123'
pattern = '[A-Z0-9][A-Z]{2} [0-9][A-Z0-9]{2,3}'

re.findall(pattern = pattern, string = words)
# repetition only applies to characters attached with curly brackets
# [A-Z]{2} --> means capital letters are repeated 2 times
# [A-Z0-9]{2,3} --> means capital letters or numbers are repeated at least 2 times and a maximum of 3 times
```

- create a pattern that matches `XRA 000, 1AA 2CD, XYZ 4810`, and `ABC 77`

```python
words = 'XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, XR1 000, ABC 77, COD 1, XY@ 123'
pattern = '[A-Z0-9][A-Z]{2} [0-9][A-Z0-9]{1,3}'

re.findall(pattern = pattern, string = words)

# pattern '[A-Z0-9][A-Z]{2} [0-9][A-Z0-9]{1,3}' is equivalent to:
# '[A-Z0-9][A-Z][A-Z] [0-9][A-Z0-9]'
# '[A-Z0-9][A-Z][A-Z] [0-9][A-Z0-9][A-Z0-9]'
# '[A-Z0-9][A-Z][A-Z] [0-9][A-Z0-9][A-Z0-9][A-Z0-9]'
```

## **8. Metacharacters**

We can use special escape codes to find certain types of patterns in your data, such as digits, non-digits, spaces, and more. For example:

Code | Meaning
--- | ---
\d | a digit [0-9]
\D | a non-digit
\w | alphanumeric [a-zA-Z0-9_]
\W | non-alphanumeric
\s | whitespace (tab, space, newline, etc.)
\S | non-whitespace

- create a pattern that matches `XRA 000, 1AA 2CD, XYZ 4810`, and `ABC 77`

```python
words = 'XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, XR1 000, ABC 77, COD 1, XY@ 123'
pattern = '[A-Z0-9][A-Z]{2} [0-9][A-Z0-9]{1,3}'

re.findall(pattern = pattern, string = words)
```

```python
words = 'XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, XR1 000, ABC 77, COD 1, XY@ 123'
pattern = '\w[A-Z]{2}\s\d\w{1,3}'

re.findall(pattern = pattern, string = words)
```

- create a pattern that matches `XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, XR1 000`, and `ABC 77`

```python
words = 'XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, XR1 000, ABC 77, COD 1, XY@ 123'
pattern = '\w{3}\s\d\w{1,3}'

re.findall(pattern = pattern, string = words)

# xyz 4810 --> join the match, because \w represents alphanumeric (uppercase, lowercase, numbers, and also underscore)
```

## **9. Exclusion**

We can use ^ to exclude terms by enclosing them in bracket syntax notation. For example: `[^...]` will match any character not in the brackets. Let's look at an example:

- create a pattern that matches `XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, ABC 77`, and `XY@ 123`

```python
words = 'XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, XR1 000, ABC 77, COD 1, XY@ 123'
pattern = '\w{2}[^\d]\s\d\w{1,3}'

re.findall(pattern = pattern, string = words)
```

## **10. Repetition Metacharacters**

In addition to the metacharacters mentioned above, there are repetition metacharacters that request more than one occurrence of a character. The following table shows all repetition metacharacters.

| Quantifier | Role |
| --- | --- |
| \* | 0 or more. Same as { , } |
| \+ | 1 or more. Same as {1, } |
| \? | Optional, so 0 or 1. Same as { ,1}. |

- `\*` matches 0 or more repetitions

```python
words = 'XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, XR1 000, ABC 77, COD 1, XY@ 123'
pattern = '\w*\s\d*'

re.findall(pattern = pattern, string = words)

# \w* --> lowercase or uppercase or number or underscore repeated 0 times or more
# \s --> space
# \d* --> number repeated 0 times or more
```

- `\+` matches 1 or more repetitions

```python
words = 'XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, XR1 000, ABC 77, COD 1, XY@ 123'
pattern = '\w+\s\d+'

re.findall(pattern = pattern, string = words)

# \w+ --> lowercase or uppercase letters or numbers or underscores repeated 1 time or more
# \s --> space
# \d+ --> numbers repeated 1 time or more

# means there is a space on the left of which is a letter or number or underscore is a number at least 1 time
# and on the right is a number at least 1 time or more
```

- `\?` matches 0 or 1 repetitions

```python
words = 'XRA 000, 1AA 2CD, xyz 4810, XYZ 4810, XR1 000, ABC 77, COD 1, XY@ 123'
pattern = '\w?\s\d?'

re.findall(pattern = pattern, string = words)
```

## **11. Grouping**

We can use grouping to search for characters that are exactly the same as the regex that is in the brackets `()`. The benefits of grouping include being able to apply loops to groups and making the regex easier to read.

```python
words = 'ab, abc, abcc, abcccc, abcabc, abcabcabc'
pattern = 'abc+'

re.findall(pattern = pattern, string = words)
```

In the example above, the repetition metacharacter '+' is attached to the character `c`, so that `c` is at least 1 time or more.

```python
words = 'ab, abc, abcc, abcccc, abcabc, abcabcabc'
pattern = '(abc)+'

re.findall(pattern = pattern, string = words)
```

In the example above, the repetition metacharacter '+' is attached to the character `abc`, so that `abc` is at least 1 time or more.
The subtring displayed is the pattern in parentheses.

```python
words = 'ab, abc, abcc, abcccc, abcabc, abcabcabc'
pattern = '(?:abc)+'

re.findall(pattern = pattern, string = words)
```

By using the `?:` sign in the grouping, all substrings that contain `abc` 1 or more times are displayed: `abc`, `abcabc`, etc.

## **12. Alternative**

We can use the `|` sign as an alternative. Its function is similar to OR in python.

- create a pattern that displays `pineapple juice` and `watermelon juice` without grouping.

```python
words = 'apple juice, melon juice, watermelon juice, apple pie, pineapple juice'
pattern = 'pineapple juice|watermelon juice'

re.findall(pattern = pattern, string = words)
```

- create a pattern that displays `pineapple juice` and `watermelon juice` with grouping.

```python
words = 'apple juice, melon juice, watermelon juice, apple pie, pineapple juice'
pattern = '(pineapple|watermelon) juice'

re.findall(pattern = pattern, string = words)
# here match with pineapple juice and watermelon juice
# but only the ones in brackets ('pineapple' and 'watermelon') are displayed
```

```python
words = 'apple juice, melon juice, watermelon juice, apple pie, pineapple juice'
pattern = '(?:pineapple|watermelon) juice'

re.findall(pattern = pattern, string = words)
# here match with pineapple juice and watermelon juice
# and what is displayed is according to pineapple juice and watermelon juice
```

## **13. Word Boundaries**

By using `\b` at the beginning and end of the pattern, we ensure that the match starts and ends as a whole word, not part of another word.

- create a pattern to display `red` from the words **red, hundred, and reddit**

```python
words = 'red, hundred, reddit'
pattern = 'red'

re.findall(pattern = pattern, string = words)
```

- create a pattern to display `red` from the words **red and reddit**

```python
# display 'red' from 'red' and 'reddit'
words = 'red, hundred, reddit'
pattern = '\\bred'

re.findall(pattern = pattern, string = words)

# output : ['red', 'red'] --> first red from red and second red from reddit
# red in hundred is not included, because before red there is a character
```

Notes: we have to use the additional escape character '\' so that Python doesn't mistake `\b` as backspace.

- create a pattern to display `red` from the words **red and hundred**

```python
words = 'red, hundred, reddit'
pattern = 'red\\b'

re.findall(pattern = pattern, string = words)

# output : ['red', 'red'] --> first red from red and second red from hundred
# red in reddit is not included, because after red there is a character
```

- create a pattern to display `red` from the word **red**

```python
words = 'red, hundred, reddit'
pattern = '\\bred\\b'

re.findall(pattern = pattern, string = words)

# output : ['red'] --> red from red
# red in hundred is not included, because before red there is a character
# red in reddit is not included, because after red there is a character
```

- create a pattern that displays `apple juice` and `melon juice`

```python
words = 'apple juice, melon juice, watermelon juice, apple pie, pineapple juice'
pattern = '(?:\\bapple|\\bmelon) juice'

re.findall(pattern = pattern, string = words)
```
