# **Regular Expression**
---

```python
# Importing regex library
import re
```

```python
# Importing request to access file
import requests

# Loading text
url = 'https://raw.githubusercontent.com/harishmuh/Python-for-Data-Science-Analysis/refs/heads/main/datasets/mbox-short.txt'
response = requests.get(url)
text = response.text
```

**1. Display IP Address**

Example:
- '141.211.14.90'
- '141.211.14.79'
- '194.35.219.184'
- '127.0.0.1'

Answer: 189 IPs

```python
# Method 1
pattern = '[1][1-9]{2}\.\d{1,3}.\d{1,3}.\d{1,3}'
len(re.findall(pattern, text))
```

```python
# Method 2
pattern = '\d{3}(?:\.\d{1,3}){3}'
len(re.findall(pattern, text))
```

**2. Show email addresses**

Examples:
- 'stephen.marquard@uct.ac.za'
- 'postmaster@collab.sakaiproject.org'
- '200801051412.m05ECIaH010327@nakamura.uits.iupui.edu'
- 'source@collab.sakaiproject.org'

Answer: There are 305 emails

```python
pattern = '[\w.]+@[\w.]+\.\w{2,3}'
len(re.findall(pattern, text))
```

**3. Display the date and time**

Example
- '2008-01-05 09:12:07'
- '2008-01-04 18:08:50'
- '2008-01-04 16:09:01'

Answer: There are 34 dates

```python
pattern = '\d{4}-\d{2}-\d{2}\s\d{2}:\d{2}:\d{2}'
len(re.findall(pattern, text))
```

**4. Display Day and Date**

Example:
- 'Sat, 05 Jan 2008'
- 'Sat, 5 Jan 2008'
- 'Sat, 05 Jan 2008'

Answer: There are 223 dates

```python
pattern = '\w{3},\s\d{1,2}\s\w{3}\s\d{4}'
len(re.findall(pattern, text))
```

**5. Display emails preceded by Author**

Example:
- 'stephen.marquard@uct.ac.za',
- 'louis@media.berkeley.edu',
- 'zqian@umich.edu'

Answer: There are 27 emails

```python
pattern = 'Author: ([\w.]+@[\w.]+\.\w{2,3})'
len(re.findall(pattern, text))
```

**6. Display emails preceded by "From" or "To"**

Examples:
- 'source@collab.sakaiproject.org',
- 'stephen.marquard@uct.ac.za',
- 'source@collab.sakaiproject.org',
- 'louis@media.berkeley.edu'

Answer: There are 54 emails

```python
pattern = '(?:From:|To:) ([\w.]+@[\w.]+\.\w{2,3})'
len(re.findall(pattern, text))
```
