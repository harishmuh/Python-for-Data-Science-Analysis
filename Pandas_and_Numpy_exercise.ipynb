**Number 1**

Create an array with the arrange command, with the start and end values as input!
From the resulting array, create 2- and 3-dimensional versions!

```python
# Input
start = 1
finish = 28
```

```python
# Creating 1 Dimensional Array using list comperhension
array_1d = [i for i in range(start, finish)]
print("1 Dimensional Array:")
print(array_1d)
```

```python
# Creating 1 Dimensional Array using list comperhension using numpy.arange
array_1d = np.arange(1, 28)
print("1-Dimensional Array (NumPy):")
print(array_1d)
```

```python
# 2D
array_2d = array_1d.reshape(3, 9)
print("2D Array:")
print(array_2d)
```

```python
#3D
array_3d = array_1d.reshape(3, 3, 3)
print("3D Array:")
print(array_3d)
```

**Number 2**

* Create a 1-dimensional array
* Convert the 1-dimensional array into a 2-dimensional matrix!
* After converting to 2-dimensional form, reverse the order of the data in each row! For example, from 1, 2, 3 to 3, 2, 1.

```python
# Create a 1-dimensional array
array_1d = np.array([23, 19, 15, 30, 30, 47, 44, 26, 41, 5, 9, 17, 46, 19, 21, 32])

print("1-Dimensional Array:")
print(array_1d)
```

```python
# 2D
array_2d = array_1d.reshape(4, 4)
print("2D Array:")
print(array_2d)
```

```python
# Swap the positions of 1,2,3 to 3,2,1
# Reverse the order of elements in each row (reverse columns)
reversed_arr = array_2d[:, ::-1]

# Display the result
print("Reversed array")
print(reversed_arr)
```

**No. 3**

* Create a 2-dimensional array with dimensions of 10 x 10!
* The contents of the array are:
  * The outer edge is the number 4.
  * Every other edge is 0.

```python
# Create a 10x10 array containing all 0s
array = np.zeros((10, 10), dtype=int)

# Change the outer values to 4
array[0, :] = 4 # top row
array[-1, :] = 4 # bottom row
array[:, 0] = 4 # left column
array[:, -1] = 4 # right column

print(array)
```

**Number 5**

* Create a series using 100 random integer numbers from 1 to 5!
* Calculate the frequency of each number!
* Find the two values with the highest frequency!

```python
# Create a Series with 100 random numbers between 1 and 5
random_series = pd.Series(np.random.randint(1, 6, size=100))
random_series
```

```python
# Count the frequency of occurrence of each number
frequencies = random_series.value_counts()
# Display the frequency of all numbers
print("Frequencies of all values:")
print(frequencies)
```

```python
# Take the 2 values with the highest frequency
top_2 = frequencies.head(2)
print("\nTop 2 most frequent values:")
print(top_2)
```

**Number 9**

* Import the "seaborn" library, then load the data named "MPG"!
* Create a new column named 'horse_origin'!
* This 'horse_origin' column must be populated using the contains function:
* If the 'horsepower' value is <105, display 'lower' + 'origin name' in the same row. For example, 'lower USA'.
* If the 'horsepower' value is >= 105, display 'upper' + 'origin name'.

```python
# Create a regular function
def classify_horse_origin(row):
  if row['horsepower'] < 105:
    return 'lower ' + row['origin'].title()
  else:
    return 'upper ' + row['origin'].title()

# Apply the function to each row of the dataframe
mpg['horse_origin'] = mpg.apply(classify_horse_origin, axis=1)

# Display the results
mpg.head()
```

**Number 10**

* Import the "seaborn" library, then load the data named "MPG"!
* Group the "weight" column based on the "origin" column by aggregating the mean, standard deviation, sum, count, min, and max!
* Reset the index so that the "origin" column is no longer an index but remains a column!

```python
# Grouping and aggregation
grouped_weight = mpg.groupby('origin')['weight'].agg(['mean', 'std', 'sum', 'count', 'min', 'max'])
grouped_weight
```

```python
# Grouping and aggregation (after resetting index())
reset = grouped_weight.reset_index()
reset
```
