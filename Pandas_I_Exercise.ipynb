**Step 1. Import the required libraries.**

**Step 2. Create a dictionary to create a DataFrame with 5 keys and each key has a value in the form of a list with 10 items. The contents of the dictionary follow the output of Step 3.**

```python
dict_employees = {
'Name' : ['Andi', 'Budi', 'Caca', 'Dedi', 'Erika', 'Fahri', 'Gina', 'Hani', 'Irja', 'Jarot'],
'Age' : [24, 25, 30, 32, 28, 24, 29, 31, 30, 29],
'Gender' : ['Male', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Female', 'Male', 'Male'],
'Occupation' : ['Data Analyst', 'Data Engineer', 'Data Scientist', 'Data Scientist', 'Data Analyst', 'Data Engineer', 'ML Engineer', 'ML Researcher', 'BI Analyst', 'Data Analyst'],
'Domicile' : ['Jakarta', 'South Tangerang', 'Jakarta', 'Bandung', 'Jakarta', 'Surabaya', 'Bandung', 'Yogyakarta', 'Jakarta', 'South Tangerang']
}
```

**Step 3. Convert the dictionary from Step.2 into DataFrame form and save it into a new variable named df.**

```python
df = pd.DataFrame(data=dict_employees)
df
```

**Step 4. Change the order of the columns to be as follows: Name, Gender, Age, Domicile, Occupation.**

```python
df = df[['Name', 'Gender', 'Age', 'Domicile', 'Occupation']].copy()
df
```

**Step 5. Add 1 new column named 'Hobby' and fill in the data as shown in the following output.**

```python
df['Hobby'] = ['Gaming', 'Futsal', 'Read', 'Gaming', 'Cooking', 'Traveling', 'Basketball', 'Traveling', 'Gaming', 'Futsal']
df
```

```python
df[df['Age']>30]
```

**10. Who is under 30 years old and whose hobby is gaming?**

```python
df.loc[(df['Age']<30) & (df['Hobby']=='Gaming')]
```

**11. Tampilkan nama orang yang pekerjaannya adalah Data Analyst! Tampilkan kolom nama saja dalam bentuk Series!**

```python
df.loc[
df['Age']>df[df['Name']=='Erika']['Age'].values[0],
['Name']
]
```

```python
df[(df['Gender']=='Female') & (df['Hobby']=='Traveling')]
```
